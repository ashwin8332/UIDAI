## Problem Statement & Approach
Analyze Aadhaar demographic update patterns to identify temporal peaks, geographic hotspots, age group trends, and anomalies. The approach combines data cleaning, temporal aggregation, geographic summaries, and statistical anomaly checks with clear visualizations.

## Datasets Used (Demographic Updates)
- **Files**: 5 CSV files (~2M records total)
- **Columns used**: `state`, `district`, `date`, `pincode`, `demo_age_5_17`, `demo_age_17_`

## Methodology (Summary)
1. Loaded and merged 5 demographic CSV files
2. Standardized state names using regex cleaning and mapping dictionary
3. Parsed dates and created `month`, `year_month` columns
4. Derived `total_updates = demo_age_5_17 + demo_age_17_`
5. Cleaned Bihar districts with specific mapping
6. Identified pincodes spanning multiple districts (flagged for review)
7. Aggregated by state/district/pincode/month for trend analysis

## Key Findings (Aligned to Observations)
- **Temporal Trends**: Update requests show monthly variations; certain months have higher activity.
- **Geographic Patterns**: High-population states (UP, Bihar, Maharashtra) dominate update volumes.
- **Age Group Insights**: Age 17+ updates significantly outweigh 5-17 age group.
- **Anomalies**: Some pincodes map to multiple districts (data quality issue flagged).

```python
# Demographic analysis prep
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

_df = df.copy()
_df['date_parsed'] = pd.to_datetime(_df['date'], errors='coerce')
_df['year_month'] = _df['date_parsed'].dt.to_period('M')
_df['month'] = _df['date_parsed'].dt.month
_df['total_updates'] = _df['demo_age_5_17'].fillna(0) + _df['demo_age_17_'].fillna(0)

# Aggregations
monthly_updates = _df.groupby('year_month')['total_updates'].sum().reset_index()
monthly_updates['year_month_str'] = monthly_updates['year_month'].astype(str)
state_updates = _df.groupby('state_clean')['total_updates'].sum().sort_values(ascending=False)
heatmap_updates = _df.pivot_table(index='state_clean', columns='month', values='total_updates', aggfunc='sum').fillna(0)

# 1) Line chart: Monthly trend
plt.figure(figsize=(14, 5))
plt.plot(monthly_updates['year_month_str'], monthly_updates['total_updates'], marker='o')
plt.title('Demographic Updates Over Time')
plt.xlabel('Month')
plt.ylabel('Total Updates')
plt.xticks(rotation=45, ha='right')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 2) Bar chart: Top 10 states
plt.figure(figsize=(10, 6))
state_updates.head(10).sort_values().plot(kind='barh', color='steelblue')
plt.title('Top 10 States by Demographic Updates')
plt.xlabel('Total Updates')
plt.ylabel('State')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 3) Heatmap: State vs Month
plt.figure(figsize=(12, 8))
_top_states = state_updates.head(15).index
sns.heatmap(heatmap_updates.loc[_top_states], cmap='YlOrRd', annot=False)
plt.title('State-Month Heatmap of Demographic Updates (Top 15 States)')
plt.xlabel('Month')
plt.ylabel('State')
plt.tight_layout()
plt.show()

# 4) Pie chart: Age group distribution
plt.figure(figsize=(6, 6))
age_5_17_total = _df['demo_age_5_17'].sum()
age_17_plus_total = _df['demo_age_17_'].sum()
plt.pie([age_5_17_total, age_17_plus_total], labels=['Age 5-17', 'Age 17+'], autopct='%1.1f%%', startangle=90, colors=['#66b3ff', '#ff9999'])
plt.title('Age Group Distribution of Demographic Updates')
plt.show()
```

```python
# Create comparison dataframe
comparison_demo = pd.DataFrame({
    'State': ['Bihar', 'Karnataka'],
    'Total Updates': [df_bihar_cleaned['total_enroll'].sum(), df_karnataka_cleaned['total_enroll'].sum()],
    'Age 5-17': [df_bihar_cleaned['demo_age_5_17'].sum(), df_karnataka_cleaned['demo_age_5_17'].sum()],
    'Age 17+': [df_bihar_cleaned['demo_age_17_'].sum(), df_karnataka_cleaned['demo_age_17_'].sum()],
    'Districts': [df_bihar_cleaned['district_clean'].nunique(), df_karnataka_cleaned['district_clean'].nunique()],
    'Pincodes': [df_bihar_cleaned['pincode'].nunique(), df_karnataka_cleaned['pincode'].nunique()],
    'Multi-District Pincodes': [df_bihar_cleaned['pin_multi_district_flag'].sum(), df_karnataka_cleaned['pin_multi_district_flag'].sum()]
})
comparison_demo
```

```python
# Comparison visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. Total Updates Comparison
axes[0, 0].bar(comparison_demo['State'], comparison_demo['Total Updates'], color=['#FF6B6B', '#9C27B0'])
axes[0, 0].set_title('Total Demographic Updates', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Total Updates')
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Age Group Comparison
x = np.arange(len(comparison_demo['State']))
width = 0.35
axes[0, 1].bar(x - width/2, comparison_demo['Age 5-17'], width, label='Age 5-17', color='#66B2FF')
axes[0, 1].bar(x + width/2, comparison_demo['Age 17+'], width, label='Age 17+', color='#FFA726')
axes[0, 1].set_title('Age Group Distribution', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Updates')
axes[0, 1].set_xticks(x)
axes[0, 1].set_xticklabels(comparison_demo['State'])
axes[0, 1].legend()
axes[0, 1].grid(axis='y', alpha=0.3)

# 3. Geographic Coverage
axes[1, 0].bar(comparison_demo['State'], comparison_demo['Districts'], color=['#FF6B6B', '#9C27B0'])
axes[1, 0].set_title('Number of Districts', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Districts')
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. Pincode Coverage
axes[1, 1].bar(comparison_demo['State'], comparison_demo['Pincodes'], color=['#FF6B6B', '#9C27B0'])
axes[1, 1].set_title('Number of Unique Pincodes', fontsize=12, fontweight='bold')
axes[1, 1].set_ylabel('Pincodes')
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
