## Problem Statement & Approach
Analyze Aadhaar enrollment patterns to detect temporal peaks, geographic hotspots, demographic distribution, and anomalies. The approach combines cleaning, temporal aggregation, geographic summaries, and visual analytics.

## Datasets Used (Enrollment)
Columns used: `state`, `district`, `date`, `pincode`, `age_0_5`, `age_5_17`, `age_18_greater`.

## Methodology (Summary)
- Standardized state/district names
- Parsed dates and created `month`, `year_month`
- Derived `total_enroll`
- Aggregated by state and month for trends

## Key Findings (Aligned to Observations)
- **Temporal Trends**: Enrollment peaks align with government drives and welfare rollouts.
- **Geographic Patterns**: Urbanized states show higher enrollment volumes; migration-heavy states show consistent activity.
- **Demographic Insights**: Highest enrollment concentration in working-age group (18+).
- **Anomalies**: Certain districts show unusually high enrollment indicating correction drives or saturation.

```python
# Enrollment analysis prep
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

_df = df.copy()
_df['date_parsed'] = pd.to_datetime(_df['date'], errors='coerce')
_df['year_month'] = _df['date_parsed'].dt.to_period('M')
_df['month'] = _df['date_parsed'].dt.month
_df['total_enroll'] = _df['age_0_5'].fillna(0) + _df['age_5_17'].fillna(0) + _df['age_18_greater'].fillna(0)

# Aggregations
monthly_enroll = _df.groupby('year_month')['total_enroll'].sum().reset_index()
monthly_enroll['year_month_str'] = monthly_enroll['year_month'].astype(str)
state_enroll = _df.groupby('state_clean')['total_enroll'].sum().sort_values(ascending=False)
heatmap_enroll = _df.pivot_table(index='state_clean', columns='month', values='total_enroll', aggfunc='sum').fillna(0)

# 1) Line chart: Monthly trend
plt.figure(figsize=(14, 5))
plt.plot(monthly_enroll['year_month_str'], monthly_enroll['total_enroll'], marker='o')
plt.title('Enrollment Over Time')
plt.xlabel('Month')
plt.ylabel('Total Enrollments')
plt.xticks(rotation=45, ha='right')
plt.grid(alpha=0.3)
plt.show()

# 2) Bar chart: Top 10 states
plt.figure(figsize=(10, 6))
state_enroll.head(10).sort_values().plot(kind='barh')
plt.title('Top 10 States by Enrollment')
plt.xlabel('Total Enrollments')
plt.ylabel('State')
plt.grid(axis='x', alpha=0.3)
plt.show()

# 3) Heatmap: State vs Month
plt.figure(figsize=(12, 8))
_top_states = state_enroll.head(15).index
sns.heatmap(heatmap_enroll.loc[_top_states], cmap='YlOrRd')
plt.title('State-Month Heatmap of Enrollment (Top 15 States)')
plt.xlabel('Month')
plt.ylabel('State')
plt.show()

# 4) Pie chart: Age-group share
plt.figure(figsize=(6, 6))
plt.pie([
    _df['age_0_5'].sum(),
    _df['age_5_17'].sum(),
    _df['age_18_greater'].sum()
], labels=['Age 0-5', 'Age 5-17', 'Age 18+'], autopct='%1.1f%%', startangle=90)
plt.title('Enrollment by Age Group')
plt.show()
```
